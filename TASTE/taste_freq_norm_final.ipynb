```python
import pandas as pd
from collections import defaultdict

d = pd.read_table('df_taste_total.tsv', sep='\t', on_bad_lines='skip')

df = d[['Taste_Word', 'year']].copy()
df['year'] = pd.to_numeric(df['year'], errors='coerce')

def get_century(year):
    return int(year) // 100 + 1 if pd.notnull(year) else None

df = df.dropna(subset=['Taste_Word', 'year'])

df['century'] = df['year'].apply(get_century)

word_freq_per_century = defaultdict(lambda: defaultdict(int))

for _, row in df.iterrows():
    century = row['century']
    words = str(row['Taste_Word']).lower().split('|')  # Split on pipes
    for word in words:
        word = word.strip()
        if century and word:
            word_freq_per_century[century][word] += 1

total_freq_per_century = {
    century: sum(freqs.values())
    for century, freqs in word_freq_per_century.items()
}

with open('taste_freq_norm_clean.tsv', 'w') as f:
    f.write("century\tword\tfreq_normal\n")
    for century in sorted(word_freq_per_century):
        total = total_freq_per_century[century]
        for word, count in sorted(
            word_freq_per_century[century].items(),
            key=lambda x: x[1],
            reverse=True
        ):
            freq_normal = count / total
            f.write(f"{century}\t{word}\t{freq_normal:.8f}\n")
```

```python
import matplotlib.pyplot as plt

df = pd.read_table('taste_freq_norm_clean.tsv', sep='\t')

num_top_words = 20
centuries = range(16, 21)

for century in centuries:
    df_century = df[df['century'] == century].head(num_top_words)

    df_century = df_century[
        df_century['word'].apply(lambda x: len(str(x)) >= 3)
    ].head(num_top_words)

    plt.figure(figsize=(10, 6))
    plt.barh(
        df_century['word'],
        df_century['freq_normal'],
        color='skyblue'
    )
    plt.xlabel('Frequency')
    plt.ylabel('Taste word')
    plt.title(f'Top {num_top_words} taste words_{century - 1}00')
    plt.gca().invert_yaxis()  # Display the most frequent words at the top
    plt.tight_layout()
    plt.show()
```
